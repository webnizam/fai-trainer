```python
# Function to download an image
def download_image(url, file_path):
    response = requests.get(url)
    if response.status_code == 200:
        with open(file_path, "wb") as file:
            file.write(response.content)
        print(f"Downloaded {file_path}")
    else:
        print(f"Failed to download {file_path}")
```

```python
for query in class_names:
    # Parameters for the search
    params = {
        "query": query,
        "per_page": 10,  # Number of images to download
        "client_id": UNSPLASH_ACCESS_KEY,
    }

    # Directory to save the downloaded images
    download_dir = f"datasets/{query}"
    os.makedirs(download_dir, exist_ok=True)
    
    # Search for cat images
    response = requests.get(UNSPLASH_SEARCH_URL, params=params)
    if response.status_code == 200:
        data = response.json()
        images = data["results"]

        for i, image in enumerate(images):
            image_url = image["urls"]["regular"]
            file_name = f"{query}_{i+1}.jpg"
            file_path = os.path.join(download_dir, file_name)
            download_image(image_url, file_path)
    else:
        print("Failed to search for images")
```
